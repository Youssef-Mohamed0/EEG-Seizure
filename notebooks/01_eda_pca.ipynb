```python
import sys; sys.path.insert(0, "..")          # project root, so `backend` can be imported
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
from backend.ml_pipeline import load_data_and_split, SEED
from backend.config import FEATURE_COLS

# The ONE official split (seed 42, stratified). Do NOT make your own split.
df, Xtr, Xte, yc_tr, yc_te, yr_tr, yr_te = load_data_and_split()
print(Xtr.shape, Xte.shape)                   # expected: (778, 10) (195, 10)
```

```python
df[FEATURE_COLS].describe().T
```

```python
corr = df[FEATURE_COLS].corr()
fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(corr, cmap='coolwarm', vmin=-1, vmax=1)
ax.set_xticks(range(10)); ax.set_xticklabels(FEATURE_COLS, rotation=90)
ax.set_yticks(range(10)); ax.set_yticklabels(FEATURE_COLS)
fig.colorbar(im); plt.tight_layout(); plt.show()
# TODO: note which features are strongly correlated (Weight/BMI?) and what that means for PCA
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
scaler = StandardScaler().fit(Xtr)            # fit on TRAIN only
pca_full = PCA(random_state=SEED).fit(scaler.transform(Xtr))
print(pca_full.explained_variance_ratio_.round(3))
print('first 3 components:', pca_full.explained_variance_ratio_[:3].sum().round(4))   # reference: 0.5275
```
