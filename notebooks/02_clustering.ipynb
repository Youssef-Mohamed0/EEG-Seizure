```python
import sys; sys.path.insert(0, "..")          # project root, so `backend` can be imported
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
from backend.ml_pipeline import load_data_and_split, SEED
from backend.config import FEATURE_COLS

# The ONE official split (seed 42, stratified). Do NOT make your own split.
df, Xtr, Xte, yc_tr, yc_te, yr_tr, yr_te = load_data_and_split()
print(Xtr.shape, Xte.shape)                   # expected: (778, 10) (195, 10)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
scaler = StandardScaler().fit(Xtr)
Xs = scaler.transform(Xtr)
for k in range(2, 7):
    km = KMeans(k, random_state=SEED, n_init=10).fit(Xs)
    print(k, 'inertia', round(km.inertia_), 'silhouette', round(silhouette_score(Xs, km.labels_), 3))   # reference: k=2 0.188, k=3 0.184
```

```python
from backend.cluster_naming import describe_clusters
km3 = KMeans(3, random_state=SEED, n_init=10).fit(Xs)
for k, v in describe_clusters(km3.cluster_centers_, FEATURE_COLS, np.bincount(km3.labels_)).items():
    print(k, v['size'], v['en'])
# Cluster ids are arbitrary: the NAMES come from the centers, never hard-code them.
```
